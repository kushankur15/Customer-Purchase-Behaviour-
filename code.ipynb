```python
item_frequency = df.groupby(['Item Purchased','Gender','Frequency of Purchases'])['Item Purchased'].value_counts().reset_index(name='frequency')
item_frequency.head(6)
```

```python
item_frequency.shape
```

```python
df['Frequency of Purchases'].value_counts()
```

```python
item_above_median = item_frequency[item_frequency['frequency'] >= item_frequency['frequency'].median()]
item_above_median.to_csv('item_above_median.csv')
item_below_median = item_frequency[item_frequency['frequency'] < item_frequency['frequency'].median()]
item_below_median.to_csv('item_below_median.csv')
```

```python
weekly_df = item_frequency[item_frequency['Frequency of Purchases'] == 'Weekly']

peaks_frequently_bought_items_weekly = weekly_df[
    (weekly_df['frequency'] == weekly_df['frequency'].max()) |
    (weekly_df['frequency'] == weekly_df['frequency'].min())
]
peaks_frequently_bought_items_weekly
```

```python
purchase_frequency_dataframes = []
indices = [j for j in range(0,7)]
frequency_of_purchase = df['Frequency of Purchases'].unique().tolist()
for freq,i in zip(frequency_of_purchase,indices):
    temp_df = item_frequency[item_frequency['Frequency of Purchases'] == freq]

    peaks_freq = temp_df[
        (temp_df['frequency'] == temp_df['frequency'].max()) |
        (temp_df['frequency'] == temp_df['frequency'].min())
    ]

    purchase_frequency_dataframes.append(peaks_freq)
```

```python
purchase_frequency_dataframes
```

```python
freq_df = pd.concat(
    purchase_frequency_dataframes,
    ignore_index=True
)[
    ['Frequency of Purchases', 'Item Purchased', 'Gender', 'frequency']
]

freq_df.to_csv('freq_df.csv')
```

```python
# products bought by females
female = item_frequency[item_frequency['Gender'] == 'Female']
female
```

```python
freq_bought_by_female = female[female['frequency'] >= 10]
freq_bought_by_female.to_csv('freq_female.csv')
```

```python
male = item_frequency[item_frequency['Gender'] == 'Male']
male
```

```python
freq_bought_by_male = male[male['frequency'] > male['frequency'].median()]
freq_bought_by_male.to_csv('freq_male.csv')
```
